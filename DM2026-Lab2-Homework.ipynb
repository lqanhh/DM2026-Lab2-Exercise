## 1. Model Development (10 pts Required)

### 1.1 Preprocessing Steps

The four dataset files were loaded and merged using `post_id`. The resulting dataset contained 3,360 labeled training posts and 1,440 test posts.

Text preprocessing included:
- Converting text to lowercase.
- Replacing URLs with the token "URL."
- Normalizing whitespace.
Punctuation, mentions, hashtags, stop words, and negations were retained because they may contain useful emotional information. No stemming or lemmatization was applied.

![Emotion distribution bar chart](./pics/report_figures/figure1_class_distribution.png)

The training data was imbalanced, with *joy* as the most frequent class (26.0%) and *fear* as the least frequent class (8.6%). Therefore, balanced class weights were used in Logistic Regression.

### 1.2 Feature Engineering Steps

Word-level and character-level TF-IDF features were combined using `FeatureUnion`.
- **Word n-grams:** Capture individual words and short word sequences.
- **Character n-grams:** Capture spelling variations, informal expressions, and character patterns.

They were combined into a single feature matrix following this pipeline:

![TF-IDF Pipeline](./pics/report_figures/figure2_feature_engineering.png)

The main configurations used the following parameters:
| **Parameter** | **Word-level TF-IDF** | **Character-level TF-IDF** |
| :---: | :---: | :---: |
| Analyzer | `word` | `char` |
| N-gram range (Model A) | (1, 3) | (2, 5) |
| N-gram range (Model B) | (1, 2) | (2, 5) |
| `min_df` | 2 | 2 |
| `max_df` | 0.95 | Not specified |
| `sublinear_tf` | True | True |
| `max_features` | 50,000 | 50,000 |

Several feature configurations were evaluated on the validation set. The experiments showed that combining word-level and character-level features yielded better results than using only word-level features. Furthermore, extending the range of word n-grams from (1, 2) to (1, 3) and including character n-grams starting at a length of 2 improved validation performance.

### 1.3 Explanation of Your Model

Logistic Regression and Linear SVM were initially compared. Logistic Regression with balanced class weights performed better, so it was chosen for further experiments.

The final Logistic Regression configuration used was:
| **Parameter** | **Value** |
| :---: | :---: |
| `C` | 0.3 |
| `max_iter` | 2,000 |
| `class_weight` | `balanced` |
| `random_state` | 42 |

The regularization parameter controls the strength of regularization. The value `C=0.3` was selected because it produced the best validation results of all the tested values. A stratified 80/20 train-validation split (`random_state=42`) was used, with Macro F1 as the evaluation metric.

# **Competition Code**

## 1. Preprocessing Steps

```python
import pandas as pd
import numpy as np
import json
import re

# Load datasets
DATA_PATH = "/kaggle/input/competitions/2026-dm-lab-2-private-competition"

with open(f"{DATA_PATH}/final_posts.json", "r", encoding="utf-8") as f:
    posts_data = json.load(f)

posts_df = pd.DataFrame(posts_data)
emotion_df = pd.read_csv(f"{DATA_PATH}/emotion.csv")
identification_df = pd.read_csv(
    f"{DATA_PATH}/data_identification.csv"
)

sample_submission = pd.read_csv(
    f"{DATA_PATH}/sampleSubmission.csv"
)

# Merge datasets using post_id
data_df = (
    posts_df
    .merge(emotion_df, on="post_id", how="left")
    .merge(identification_df, on="post_id", how="left")
)

# Separate training and test datasets
train_df = data_df[data_df["set"] == "train"].copy()
test_df = data_df[data_df["set"] == "test"].copy()

# Text preprocessing
def clean_text(text):
    text = str(text).lower()

    # Replace URLs with a common token
    text = re.sub(r"https?://\S+|www\.\S+", " URL ", text)

    # Normalize whitespace
    text = re.sub(r"\s+", " ", text).strip()

    return text

train_df["clean_text"] = train_df["text"].apply(clean_text)
test_df["clean_text"] = test_df["text"].apply(clean_text)

# Prepare model inputs
X_train = train_df["clean_text"]
y_train = train_df["label"]

X_test = test_df["clean_text"]

# Check dataset sizes
print("Training data:", train_df.shape)
print("Test data:", test_df.shape)

print("\nLabel distribution:")
print(y_train.value_counts())
```

## 2. Feature Engineering Steps

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.pipeline import FeatureUnion

# Function to create combined word and character TF-IDF features
def make_features(word_ngram, char_ngram):

    return FeatureUnion([
        (
            "word",
            TfidfVectorizer(
                analyzer="word",
                ngram_range=word_ngram,
                min_df=2,
                max_df=0.95,
                sublinear_tf=True,
                max_features=50000
            )
        ),
        (
            "char",
            TfidfVectorizer(
                analyzer="char",
                ngram_range=char_ngram,
                min_df=2,
                sublinear_tf=True,
                max_features=50000
            )
        )
    ])

# Model A: Word (1,3) + Character (2,5)
features_A = make_features(
    word_ngram=(1, 3),
    char_ngram=(2, 5)
)

# Model B: Word (1,2) + Character (2,5)
features_B = make_features(
    word_ngram=(1, 2),
    char_ngram=(2, 5)
)

print("Feature configurations defined successfully.")
```

## 3. Model Implementation Steps

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import f1_score

# ==========================================
# 1. Train-validation split
# ==========================================

X_tr, X_val, y_tr, y_val = train_test_split(
    X_train,
    y_train,
    test_size=0.2,
    stratify=y_train,
    random_state=42
)

# ==========================================
# 2. Train Model A
# Word (1,3) + Character (2,5)
# ==========================================

features_A_val = make_features((1, 3), (2, 5))

X_A_train = features_A_val.fit_transform(X_tr)
X_A_val = features_A_val.transform(X_val)

model_A = LogisticRegression(
    C=0.3,
    max_iter=2000,
    class_weight="balanced",
    random_state=42
)

model_A.fit(X_A_train, y_tr)

proba_A = model_A.predict_proba(X_A_val)

# ==========================================
# 3. Train Model B
# Word (1,2) + Character (2,5)
# ==========================================

features_B_val = make_features((1, 2), (2, 5))

X_B_train = features_B_val.fit_transform(X_tr)
X_B_val = features_B_val.transform(X_val)

model_B = LogisticRegression(
    C=0.3,
    max_iter=2000,
    class_weight="balanced",
    random_state=42
)

model_B.fit(X_B_train, y_tr)

proba_B = model_B.predict_proba(X_B_val)

# ==========================================
# 4. Evaluate individual models
# ==========================================

assert np.array_equal(model_A.classes_, model_B.classes_)

pred_A = model_A.classes_[np.argmax(proba_A, axis=1)]
pred_B = model_B.classes_[np.argmax(proba_B, axis=1)]

print("Model A Macro F1:",
      f1_score(y_val, pred_A, average="macro"))

print("Model B Macro F1:",
      f1_score(y_val, pred_B, average="macro"))

# ==========================================
# 5. Ensemble weight experiments
# ==========================================

for weight_A in [0.5, 0.6, 0.7, 0.8, 0.9]:

    weight_B = 1 - weight_A

    ensemble_proba = (
        weight_A * proba_A +
        weight_B * proba_B
    )

    ensemble_pred = model_A.classes_[
        np.argmax(ensemble_proba, axis=1)
    ]

    score = f1_score(
        y_val,
        ensemble_pred,
        average="macro"
    )

    print(
        f"A={weight_A:.1f}, B={weight_B:.1f}: "
        f"Macro F1 = {score:.6f}"
    )

# ==========================================
# 6. Retrain Model A on full training data
# ==========================================

features_A_full = make_features((1, 3), (2, 5))

XA_full = features_A_full.fit_transform(X_train)
XA_test = features_A_full.transform(X_test)

model_A_full = LogisticRegression(
    C=0.3,
    max_iter=2000,
    class_weight="balanced",
    random_state=42
)

model_A_full.fit(XA_full, y_train)

proba_A_test = model_A_full.predict_proba(XA_test)

# ==========================================
# 7. Retrain Model B on full training data
# ==========================================

features_B_full = make_features((1, 2), (2, 5))

XB_full = features_B_full.fit_transform(X_train)
XB_test = features_B_full.transform(X_test)

model_B_full = LogisticRegression(
    C=0.3,
    max_iter=2000,
    class_weight="balanced",
    random_state=42
)

model_B_full.fit(XB_full, y_train)

proba_B_test = model_B_full.predict_proba(XB_test)

# ==========================================
# 8. Final ensemble: 60% Model A + 40% Model B
# ==========================================

assert np.array_equal(
    model_A_full.classes_,
    model_B_full.classes_
)

ensemble_test = (
    0.6 * proba_A_test +
    0.4 * proba_B_test
)

ensemble_predictions = model_A_full.classes_[
    np.argmax(ensemble_test, axis=1)
]

# ==========================================
# 9. Generate final submission
# ==========================================

submission = pd.DataFrame({
    "id": test_df["post_id"].values,
    "label": ensemble_predictions
})

# Validate submission
assert submission.shape == (1440, 2)
assert submission.isna().sum().sum() == 0
assert set(submission["label"].unique()).issubset(
    set(y_train.unique())
)

submission.to_csv(
    "submission_ensemble_60_40.csv",
    index=False
)

print("\nFinal submission:", submission.shape)
print("\nPrediction distribution:")
print(submission["label"].value_counts())

print("\nSubmission saved successfully!")
```
